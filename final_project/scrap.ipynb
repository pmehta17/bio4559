```python
import matplotlib.pyplot as plt

def plot_infection_rates(data):
    """
    data: dict where keys are age groups (str)
          and values are lists of mean infection rates (length 366).
    """
    days = list(range(366))  # 0–365

    plt.figure()

    for age_group, means in data.items():
        plt.plot(days, means, label=age_group)

    plt.xlabel("Day")
    plt.ylabel("Mean Infection Rate")
    plt.title("Infection Rate Over 365 Days by Age Group")
    plt.legend()
    plt.tight_layout()
    plt.show()


import numpy as np

data = {
    "18–25": np.linspace(0.01, 0.15, 366),
    "26–40": np.linspace(0.02, 0.10, 366),
    "40+":   np.linspace(0.015, 0.09, 366)
}

plot_infection_rates(data)

```
